```python
def morphology_function(
    input_folder,
    output_folder=r"C:\Users\arafa\Desktop\2403_Project_Work\Dataset\Morph_Frames"
):

    os.makedirs(output_folder, exist_ok=True)

    kernel = np.ones((3,3), np.uint8)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".png"):

            path = os.path.join(input_folder, filename)
            img = cv2.imread(path, 0)

            if img is None:
                continue

            #VERY LIGHT processing only
            morphed = cv2.dilate(img, kernel, iterations=1)

            save_path = os.path.join(output_folder, filename)
            cv2.imwrite(save_path, morphed)

    print("Morphology (light) completed.")
    return output_folder
```
