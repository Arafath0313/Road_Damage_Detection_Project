```python
def show_combined_frames(
    original_folder,
    contrast_folder,
    segmented_folder,
    delay=30
):

    original_files = sorted(os.listdir(original_folder))
    contrast_files = sorted(os.listdir(contrast_folder))
    segmented_files = sorted(os.listdir(segmented_folder))

    total_frames = min(len(original_files), len(contrast_files), len(segmented_files))

    for i in range(total_frames):

        orig = cv2.imread(os.path.join(original_folder, original_files[i]))
        cont = cv2.imread(os.path.join(contrast_folder, contrast_files[i]))
        seg  = cv2.imread(os.path.join(segmented_folder, segmented_files[i]))

        if orig is None or cont is None or seg is None:
            continue

        
        h, w = 480, 640
        orig = cv2.resize(orig, (w, h))
        cont = cv2.resize(cont, (w, h))
        seg  = cv2.resize(seg,  (w, h))

        
        combined = cv2.hconcat([orig, cont, seg])

        
        screen_width = 1400
        scale = screen_width / combined.shape[1]
        new_width = int(combined.shape[1] * scale)
        new_height = int(combined.shape[0] * scale)

        combined = cv2.resize(combined, (new_width, new_height))

        cv2.imshow("Pipeline View", combined)

        if cv2.waitKey(delay) & 0xFF == ord('q'):
            break

    cv2.destroyAllWindows()
```
