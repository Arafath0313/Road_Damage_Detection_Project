```python
def contour_function(
    input_folder,
    output_folder=r"C:\Users\arafa\Desktop\2403_Project_Work\Dataset\Contour_Frames",
    min_area=200 
):

    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".png"):

            path = os.path.join(input_folder, filename)
            img = cv2.imread(path, 0)

            if img is None:
                continue

            
            contours, _ = cv2.findContours(
                img,
                cv2.RETR_EXTERNAL,
                cv2.CHAIN_APPROX_SIMPLE
            )

            
            clean = np.zeros_like(img)

            count = 0  

            for cnt in contours:
                area = cv2.contourArea(cnt)

                
                if area > min_area:
                    cv2.drawContours(clean, [cnt], -1, 255, thickness=cv2.FILLED)
                    count += 1

            
            save_path = os.path.join(output_folder, filename)
            cv2.imwrite(save_path, clean)

            

    print(" Contour filtering + segmentation completed.")
    return output_folder
```
