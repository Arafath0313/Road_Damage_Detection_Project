```python
def final_overlay_function(
    contour_folder,
    gray_folder,
    output_folder=r"C:\Users\arafa\Desktop\2403_Project_Work\Dataset\Segmented_Frames",
    min_area=200
):

    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(contour_folder):
        if filename.lower().endswith(".png"):

            contour_path = os.path.join(contour_folder, filename)
            gray_path = os.path.join(gray_folder, filename)

            contour_img = cv2.imread(contour_path, 0)
            gray_img = cv2.imread(gray_path)

            if contour_img is None or gray_img is None:
                continue

            output_img = gray_img.copy()

            
            contours, _ = cv2.findContours(
                contour_img,
                cv2.RETR_EXTERNAL,
                cv2.CHAIN_APPROX_SIMPLE
            )

            
            img_h = contour_img.shape[0]

            for cnt in contours:
                area = cv2.contourArea(cnt)

                if area > min_area:
                    x, y, w, h = cv2.boundingRect(cnt)

                   
                    if y < img_h * 0.05 or y > img_h * 0.50:
                        continue

                    
                    cv2.rectangle(output_img, (x, y), (x+w, y+h), (0, 0, 255), 2)

            save_path = os.path.join(output_folder, filename)
            cv2.imwrite(save_path, output_img)

            

    print("✅ Overlay detection (filtered borders) completed.")
    return output_folder
```
